```python
import numpy as np
#Create an array [10, 20, 30, 40, 50]. Use np.where() to find the indexes of elements greater than 25.
arr=np.array([10, 20, 30, 40, 50])
index=np.where(arr>25)
index
```

```python
#Create an array [5, 10, 15, 20, 25]. Find the indexes of elements less than 15.
arr=np.array([5, 10, 15, 20, 25])
index=np.where(arr<15)
index
```

```python
#Create an array [10, 20, 30, 40, 50]. Find the indexes of elements equal to 30.
arr=np.array([10, 20, 30, 40, 50])
index=np.where(arr==30)
index
```

```python
#Create an array from 1 to 20. Find the indexes of all even numbers.
arr=np.arange(1,21)
print(arr)
even=np.where(arr%2==0)
even
```

```python
#Create an array from 1 to 20. Find the indexes of all odd numbers.
arr=np.arange(1,21)
odd=np.where(arr%2==1)
odd
```

```python
#Create an array from 1 to 10. Replace all even numbers with 0 and odd numbers with 1 using np.where().
arr=np.arange(1,11)
even_odd=np.where(arr%2==0,0,1)
even_odd
```

```python
#Create an array [10, 20, 30, 40, 50]. Replace values greater than 30 with 100, otherwise keep the original value.
arr=np.array([10, 20, 30, 40, 50])
greater=np.where(arr>30,100,arr)
greater
```

```python
#Create an array [5, 15, 25, 35, 45]. Replace values less than 25 with -1, otherwise keep the original value.
arr=np.array([5, 15, 25, 35, 45])
lessthen=np.where(arr<25,-1,arr)
lessthen
```

```python
#Create an array from 1 to 20. Convert:
#Even → 100
#Odd → 200
arr=np.arange(1,21)
even_odd=np.where(arr%2==0,100,200)
even_odd
```

```python
#Create an array [10, 25, 40, 55, 70]. Replace values greater than or equal to 40 with 1, otherwise 0
arr=np.array([10, 25, 40, 55, 70])
equal=np.where(arr>=40,1,0)
equal
```

```python
#Create an array from 1 to 30. Using np.where(), replace numbers between 10 and 20 with 100.
arr=np.arange(1,31)
replace=np.where((arr>=10) & (arr<=20),100,arr)
print(replace)
```

```python
#Create an array from 1 to 50. Replace numbers divisible by 5 with 0, otherwise keep the original values.
arr=np.arange(1,51)
replace=np.where(arr%5==0,0,arr)
replace
```

```python
#Create an array from 1 to 30. Replace numbers that are both even and greater than 10 with -1.
arr=np.arange(1,31)
replace=np.where((arr%2==0) & (arr>10),-1,arr)
replace
```

```python
#Create an array from 1 to 50. Replace numbers that are divisible by 3 OR 5 with 0.
arr=np.arange(1,51)
replace=np.where((arr%3==0) & (arr%5==0),0,arr)
replace
```

```python
#Replace numbers greater than 20 AND less than 40 with 999.
arr=np.array([10, 15, 20, 25, 30, 35, 40])
replace=np.where((arr>20) & (arr<40),999,arr)
replace
```

```python
#Using the same array, replace all values greater than 50 with 0.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
replace=np.where(arr>50,0,arr)
replace
```

```python
#Using the same array, replace all values less than or equal to 40 with -1.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
replace=np.where(arr<=40,-1,arr)
replace
```

```python
#Create a 4×4 array containing numbers from 1 to 16. Replace all even numbers with 0 and odd numbers with 1.
arr=np.arange(1,17).reshape(4,4)
print(arr)
replace=np.where(arr%2==0,0,1)
replace
```

```python
#Create a 3×4 array containing numbers from 1 to 12. Find the row and column positions of values greater than 7.
arr=np.arange(1,13).reshape(3,4)
print(arr)
position=np.where(arr>7)
position
```

```python
#Generate 10 random integers between 1 and 100. Using np.where(), replace:
#< 50 → 0
#>= 50 → 1
arr=np.random.randint(1,101,10)
print(arr)
replace=np.where(arr<50,0,np.where(arr>=50,1,arr))
replace
```

```python
#Generate a 5×5 random integer array between 1 and 100. Replace values greater than 80 with 100, otherwise keep the original values.
arr=np.random.randint(1,101,(5,5))
print(arr)
replace=np.where(arr>80,100,arr)
replace
```

```python
#Generate 20 random integers between 1 and 100. Use np.where() to classify them as:
#< 30 → "Low"
#30–70 → "Medium"
#> 70 → "High"
arr=np.random.randint(1,101,20)
print(arr)
classify=np.where(arr<30,"Low",np.where((arr>=30) & (arr<=70),"Medium",np.where(arr>70,"High",arr)))
classify
```
