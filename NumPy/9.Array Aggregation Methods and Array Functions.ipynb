```python
import numpy as np
```

```python
#Create an array [10, 20, 30, 40, 50] and find its sum using .sum().
arr=np.array([10, 20, 30, 40, 50])
arr.sum()
```

```python
#Find the minimum value of:
arr=np.array([25, 10, 45, 5, 30])
arr.min()
```

```python
#Find the maximum value of:
arr=np.array([25, 10, 45, 5, 30])
arr.max()
```

```python
#Find the mean (average) of:
arr=np.array([10, 20, 30, 40, 50])
arr.mean()
```

```python
#Find the median of:
arr=np.array([10, 20, 30, 40, 50])
np.median(arr)
```

```python
#Find the standard deviation of:
arr=np.array([10, 20, 30, 40, 50])
arr.std()
```

```python
#Create an array from 1 to 100 and find its:
#Sum
#Minimum
#Maximum
#Mean
arr=np.arange(1,101)
print(arr.sum())
print(arr.min())
print(arr.max())
print(arr.mean())
```

```python
#Find the total sum.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum()
```

```python
#Find the minimum value in the same 2D array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.min()
```

```python
#Find the maximum value in the same 2D array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.max()
```

```python
#Find the mean of the entire array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean()
```

```python
#Find the sum of each row using axis=1.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum(axis=1)
```

```python
#Find the sum of each column using axis=0.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum(axis=0)
```

```python
#Find the maximum value from each row.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.max(axis=1)
```

```python
#Find the minimum value from each column.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.min(axis=0)
```

```python
#Find the mean of each row.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean(axis=1)
```

```python
#Find the mean of each column.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean(axis=0)
```
