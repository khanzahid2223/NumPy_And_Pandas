```python
import numpy as np
```

```python
#Create an array [10, 20, 30, 40, 50] and find its sum using .sum().
arr=np.array([10, 20, 30, 40, 50])
arr.sum()
```

```python
#Find the minimum value of:
arr=np.array([25, 10, 45, 5, 30])
arr.min()
```

```python
#Find the maximum value of:
arr=np.array([25, 10, 45, 5, 30])
arr.max()
```

```python
#Find the mean (average) of:
arr=np.array([10, 20, 30, 40, 50])
arr.mean()
```

```python
#Find the median of:
arr=np.array([10, 20, 30, 40, 50])
np.median(arr)
```

```python
#Find the standard deviation of:
arr=np.array([10, 20, 30, 40, 50])
arr.std()
```

```python
#Create an array from 1 to 100 and find its:
#Sum
#Minimum
#Maximum
#Mean
arr=np.arange(1,101)
print(arr.sum())
print(arr.min())
print(arr.max())
print(arr.mean())
```

```python
#Find the total sum.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum()
```

```python
#Find the minimum value in the same 2D array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.min()
```

```python
#Find the maximum value in the same 2D array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.max()
```

```python
#Find the mean of the entire array.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean()
```

```python
#Find the sum of each row using axis=1.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum(axis=1)
```

```python
#Find the sum of each column using axis=0.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.sum(axis=0)
```

```python
#Find the maximum value from each row.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.max(axis=1)
```

```python
#Find the minimum value from each column.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.min(axis=0)
```

```python
#Find the mean of each row.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean(axis=1)
```

```python
#Find the mean of each column.
arr = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])
arr.mean(axis=0)
```

```python
#Find the index of the maximum value using np.argmax().
arr = np.array([10, 20, 30, 40, 50])
arr.argmax()
```

```python
#Find the index of the minimum value using np.argmin().
arr = np.array([10, 20, 30, 40, 50])
arr.argmin()
```

```python
#Find the index of the maximum value.
arr = np.array([10, 50, 20, 80, 30])
arr.argmin()
```

```python
#Count how many times 20 occurs.
arr=np.array([10, 20, 30, 20, 40, 20])
np.unique(arr,return_counts=True)
```

```python
#Create an array from 1 to 20 and calculate the cumulative sum using np.cumsum()
arr=np.arange(1,21)
arr.cumsum()
```

```python
#Find the square root of every element using np.sqrt().
arr = np.array([1, 4, 9, 16, 25])
np.sqrt(arr)
```

```python
#Calculate the square of every element.
arr = np.array([2, 3, 4, 5])
result=arr**2
print(result)
```

```python
#Round all values using np.round().
arr = np.array([10.5, 20.7, 30.2, 40.9])
np.round(arr)
```
