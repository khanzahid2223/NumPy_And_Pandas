```python
import numpy as np
#Sort it in ascending order using np.sort().
arr=np.array([50, 20, 40, 10, 30])
np.sort(arr)
```

```python
#Sort the array in descending order.
arr=np.array([10, 5, 8, 20, 3])
np.sort(arr)[::-1]
```

```python
#Create an array from 1 to 20 in random order. Sort it in ascending order.
arr=np.random.randint(1,21,10)
np.sort(arr)
```

```python
#Find the sorted array without changing the original array.
arr=np.array([30, 10, 50, 20, 40])
print(arr)
duplicate=np.sort(arr)
print(duplicate)
```

```python
#Sort the array and observe how duplicate values are handled.
arr=np.array([5, 2, 8, 2, 9, 5, 1])
print(arr)
arr2=np.sort(arr)
print(arr2)
np.unique(arr2,return_counts=True)
```

```python
#Sort each row in ascending order.
arr = np.array([
    [30, 10, 20],
    [60, 40, 50],
    [90, 70, 80]
])
print(np.sort(arr,axis=1))
```

```python
#Using the same array, sort each column in ascending order.
arr = np.array([
    [30, 10, 20],
    [60, 40, 50],
    [90, 70, 80]
])
print(np.sort(arr,axis=0))
```

```python
#Sort each row in ascending order.
arr = np.array([
    [50, 20, 80],
    [30, 90, 10],
    [70, 40, 60]
])
print(np.sort(arr,axis=1))
```

```python
#Sort the columns of the same array in descending order.
arr = np.array([
    [50, 20, 80],
    [30, 90, 10],
    [70, 40, 60]
])
result=np.sort(arr,axis=0)[::-1]
print(result)
```

```python
#Create a 4×5 random integer array between 1 and 100. Sort each row.
arr=np.random.randint(1,101,(4,5))
print(np.sort(arr,axis=1))
```

```python
#Find the sorted array and then find its minimum and maximum.
arr=np.array([10, 50, 20, 80, 30])
print(np.sort(arr))
print(f"Minimum:{arr.min()}")
print(f"Maximum:{arr.max()}")
```

```python
#Sort it and find the second-smallest value.
arr=np.array([40, 10, 70, 20, 90, 30])
result=np.sort(arr)
print(result)
print(f"Second Smallest Element:{result[1]}")
```

```python
#Sort it and find the second-largest value.
arr=np.array([40, 10, 70, 20, 90, 30])
result=np.sort(arr)
print(result)
print(f"Second Largest Value:{result[-2]}")
```

```python
#Create an array from 1 to 20 in random order. Sort it and calculate the median.
arr=np.random.randint(1,21,10)
result=np.sort(arr)
print(result)
print(f"Median:{np.median(result)}")
```

```python
#Generate 10 random integers between 1 and 100. Sort them and print the first 3 smallest values.
arr=np.random.randint(1,100,10)
result=np.sort(arr)
print(result)
print(f"First three smallest values:{result[:3]}")
```

```python
#Filter values greater than 30, then sort the filtered values.
arr=np.array([10, 45, 20, 80, 35, 60, 15])
greter=arr[arr>30]
print(greter)
print(np.sort(greter))
```

```python
#Create an array from 1 to 50 in random order. Filter all even numbers, then sort them.
arr=np.random.randint(1,50,10)
print(arr)
even=arr[arr%2==0]
print(even)
print(np.sort(even))
```

```python
#Generate 20 random integers between 1 and 100. Filter numbers greater than 50, then sort them in descending order.
arr=np.random.randint(1,100,20)
print(arr)
greter=arr[arr>50]
print(greter)
print(np.sort(greter)[::-1])
```
