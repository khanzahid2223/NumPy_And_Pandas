```python
import numpy as np
import pandas as pd
```

```python
#Create a Series from a list
arr=np.array([10,20,30,40,50,60])
arr_series=pd.Series(arr)
arr_series
```

```python
#Create a Pandas Series using the list [10, 20, 30, 40, 50].Show hint
arr=np.array([10, 20, 30, 40, 50])
arr_series=pd.Series(arr)
arr_series
```

```python
#Create a Series of names
arr=np.array([10, 20, 30, 40, 50])
arr_series=pd.Series(arr,name="List")
arr_series
```

```python
#Create a Series containing the names: ['Rahul', 'Aman', 'Zahid', 'Sara']
Name=np.array(['Tayab', 'Arosh', 'Zahid', 'Pathan'])
Name_Series=pd.Series(Name)
Name_Series
```

```python
#Create a Series from the tuple (100, 200, 300, 400, 500)
arr=np.array((100, 200, 300, 400, 500))
arr_series=pd.Series(arr,name="Tuple")
arr_series
```

```python
#Create a Series with values [85, 90, 78, 95] and index labels ['Maths', 'Python', 'SQL', 'AI']
values=np.array([85, 90, 78, 95])
arr_series=pd.Series(values,
                     index= ['Maths', 'Python', 'SQL', 'AI'])
arr_series
```

```python
#Create a Series from {'Apple': 100, 'Banana': 40, 'Orange': 60}
arr={'Apple': 100, 'Banana': 40, 'Orange': 60}
arr_series=pd.Series(arr)
print(arr_series)
```

```python
#Create a Series for sales over five days: [1200, 1500, 1800, 1300, 2000]. Use index labels Mon, Tue, Wed, Thu, Fri
Sales=[1200, 1500, 1800, 1300, 2000]
arr=pd.Series(Sales,
              index=["Monday","Tuesday","Wednesday","Thursday","Friday"])
arr
```

```python
#Create a Series using [10, 20, 30, 40] with the data type float
arr=pd.Series([10, 20, 30, 40],dtype="float")
arr
```

```python
#Create a Series containing five zeros, with index labels A, B, C, D, E
arr=np.zeros(5,dtype="int")
arr_series=pd.Series(arr,
                     index=["A","B","C","D","E"])
arr_series
```
