```python
import pandas as pd
```

```python
#Student Marks — Easy
#Create a Series with values [85, 90, 78, 95, 88] and indexes ['Maths', 'Python', 'SQL', 'AI', 'ML'].
#Slice from Python to AI, including both labels.
Marks=[85, 90, 78, 95, 88]
Subjects=['Maths', 'Python', 'SQL', 'AI', 'ML']
Marks_Series=pd.Series(Marks,index=Subjects)
print(Marks_Series["Python":"AI"])
```

```python
#Employee Salaries — Easy
#Create a Series with values [25000, 30000, 28000, 35000, 40000] and indexes ['E101', 'E102', 'E103', 'E104', 'E105'].
#Display salaries from E102 to E104.
Values=[25000, 30000, 28000, 35000, 40000]
Employee=['E101', 'E102', 'E103', 'E104', 'E105']
Employee_Series=pd.Series(Values,index=Employee)
Employee_Series["E102":"E104/"]
```

```python
#Product Prices — Easy
#Create a Series with values [100, 40, 60, 150, 200] and indexes ['Apple', 'Banana', 'Orange', 'Mango', 'Grapes'].
#Slice from Banana to Mango.
Price=[100, 40, 60, 150, 200]
Fruits=['Apple', 'Banana', 'Orange', 'Mango', 'Grapes']
Fruits_Series=pd.Series(Price,index=Fruits)
Fruits_Series["Banana":"Mango"]
```

```python
#Daily Sales — Easy
#Create a Series with values [1200, 1500, 1800, 1300, 2000, 2500] and indexes ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat'].
#Display sales from Tue to Fri.
Sales=[1200, 1500, 1800, 1300, 2000, 2500]
Days=['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat']
Sales_Series=pd.Series(Sales,index=Days)
Sales_Series["Tue":"Fri"]
```

```python
#Product Inventory — Challenge
#Create a Series with values [50, 20, 75, 10, 90, 40] and indexes ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Printer', 'Speaker'].
#Slice from Mouse to Printer, then change the Keyboard value to 100.
Price=[50, 20, 75, 10, 90, 40]
Prodect=['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Printer', 'Speaker']
Prodect_Series=pd.Series(Price,index=Prodect)
print(Prodect_Series["Mouse":"Printer"])
Prodect_Series["Keyboard"]=100
print(Prodect_Series)
```

```python
# Sales Analysis — Challenge
#Create a Series with values [10000, 15000, 8000, 20000, 12000, 18000, 25000] and indexes ['Store_A', 'Store_B', 'Store_C', 'Store_D', 'Store_E', 'Store_F', 'Store_G'].
# Display values from Store_B to Store_E.
# Display values from Store_D to Store_G.
# Update Store_C sales to 11000.
Price=[10000, 15000, 8000, 20000, 12000, 18000, 25000]
Store=['Store_A', 'Store_B', 'Store_C', 'Store_D', 'Store_E', 'Store_F', 'Store_G']
Store_Series=pd.Series(Price,index=Store)
print(Store_Series)
print(Store_Series["Store_B":"Store_E"])
print(Store_Series["Store_D":"Store_G"])
Store_Series["Store_C"]=11000
print(Store_Series)
```
