```python
import pandas as pd
```

```python
### Q1. Employee Salary Bonus
# → Apply a 10% bonus to every employee's salary using `apply()` and a lambda function.
# → Store the result in a new Series called `bonus_salary`.
salary = pd.Series([25000, 30000, 45000, 50000, 60000])
Bonus=salary.apply(lambda sal:sal+(sal*10/100))
Bonus
```

```python
### Q2. Product Price Discount
# → Apply a 15% discount to every product price.
# → Store the discounted prices in a new Series.
prices = pd.Series([500, 1000, 1500, 2000, 2500])


def discounted_prices(price):
    return price + (price * 15 / 100)
new=prices.apply(discounted_prices)
new
```

```python
### Q3. Student Grade Calculator
# → Assign grades using `apply()`:
# → Marks 90 or above: "A"
# → Marks 75–89: "B"
# → Marks 50–74: "C"
# → Marks below 50: "Fail"
marks = pd.Series([95, 82, 67, 45, 30, 76, 88])
def grade(mark):
    if mark>=90:
        return "A"
    elif mark>=75:
        return "B"
    elif mark>=50: 
        return "C"
    else:
        return "Fail"
new_marks=marks.apply(grade)
print(marks)
print(new_marks)
```

```python
### Q4. Employee Experience Category
# → Categorize experience using `apply()`:
# → Less than 2 years: "Fresher"
# → 2–5 years: "Mid-Level"
# → More than 5 years: "Senior"
experience = pd.Series([0, 1, 3, 5, 8, 12, 15])
def employee_experience(exp):
    if exp>=5:
        return "Senior"
    elif exp>=2:
        return "Mid-Level"
    else:
        return "Fresher"
experience.apply(employee_experience)
```

```python
### Q5. E-commerce Order Status
# → Classify orders using `apply()`:
# → Amount 1000 or above: "High Value"
# → Amount below 1000: "Regular"
orders = pd.Series([1200, 450, 800, 2500, 150, 1800])
def Ecommerce(ord):
    if ord>=1000:
        return "High Value"
    else:
        return "Regular"

orders.apply(Ecommerce)
```

```python
### Q6. Customer Age Group
# → Categorize ages using `apply()`:
# → Below 18: "Minor"
# → 18–35: "Young Adult"
# → 36–59: "Middle-Aged"
# → 60 or above: "Senior"
ages = pd.Series([16, 22, 35, 45, 60, 72, 28])
def Customer_age(age):
    if age<=18:
        return "Minor"
    elif age<=35:
        return "Young Adult"
    elif age<=59:
        return "Middle-Aged"
    else:
        return "Senior"
ages.apply(Customer_age)
```

```python
## Q7. Website Traffic Classification
# → Classify traffic using `apply()`:
# → Below 1000: "Low"
# → 1000–2999: "Medium"
# → 3000 or above: "High"
visitors = pd.Series([500, 1500, 3000, 700, 5000, 2500])
def website_trafiic(visit):
    if visit<=1000:
        return "Low"
    elif visit<2999:
        return "Medium"
    else:
        return "High"
visitors.apply(website_trafiic)
```

```python
### Q8. Employee Performance Bonus
# → Calculate a bonus percentage using `apply()`:
# → Score 90 or above: 20%
# → Score 75–89: 10%
# → Score 50–74: 5%
# → Score below 50: 0%
performance = pd.Series([45, 60, 75, 90, 95, 30, 80])
def employee_performance(perf):
    if perf>=90:
        return perf*20/100
    elif perf>=75:
        return perf*10/100
    elif perf>=50:
        return perf*5/100
    else:
        return perf*0/100
bonus=performance.apply(employee_performance)
print(bonus)
```

```python
### Q9. Transaction Fee Calculator
# → Calculate a 2% transaction fee for every transaction using `apply()`.
# → Store the fees in a new Series.
transactions = pd.Series([500, 1000, 2500, 5000, 10000])
def fee_calculation(tren):
    return tren*2/100
new_transaction=transactions.apply(fee_calculation)
print(new_transaction)
```

```python

### Q10. Text Data Cleaning
# → Remove leading and trailing spaces.
# → Convert each customer's name to title case.
# → Use `apply()` with a lambda function and string methods.
customers = pd.Series(["  zahid khan ", "ALI AHMED", "  Priya Sharma", "RAHUL KUMAR  ", " sara john "])
Cleaned=customers.apply(lambda custmer:custmer.strip().title())
Cleaned
```
