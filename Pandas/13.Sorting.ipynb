```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ]
)
```

```python
#Sort by value-low to high
salary.sort_values()
```

```python
#Sort by value-high to low
salary.sort_values(ascending=False)
```

```python
#Sort by index-A to Z
salary.sort_index()
```

```python
#Sort by index-Z to A
salary.sort_index(ascending=False)
```

```python
#Inplace Sorting
salary.sort_index(inplace=True)
salary
```

```python
#Top Values
salary.nlargest(3)
```

```python
#Bottom Values
salary.nsmallest(3)
```

```python
#Ranking
salary.rank()
```

```python
#Ranking
salary.rank(ascending=False)
```
