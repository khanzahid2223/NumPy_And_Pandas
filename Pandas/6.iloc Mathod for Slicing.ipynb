```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ]
)

print(salary)
```

```python
#Display the salaries of the first three employees.Start at 0; stop at 3.
print(salary.iloc[0:3])
```

```python
#Display the salaries of the first five employees.Stop at position 5.
print(salary.iloc[0:5])
```

```python
#Display the salaries of employees at positions 2, 3, and 4.Start at 2; stop at 5.
print(salary.iloc[2:5])
```

```python
#Display the salaries of employees at positions 4 through 7.Start at 4; stop at 8.
print(salary.iloc[4:8])
```

```python
#Display the salaries of the last three employees.Use negative slicing.
print(salary.iloc[-1:-4:-1])
```

```python
#Display all salaries except the first two employees.Start at position 2.
print(salary.iloc[2:])
```

```python
#Display all salaries except the last two employees.Use a negative stop position.
print(salary.iloc[:-2])
```

```python
#Display salaries of employees in reverse order.Use a negative step.
print(salary.iloc[::-1])
```

```python
#Display the salary of every second employee, starting with the first.Use the step argument.
print(salary.iloc[0::2])
```

```python
#Display salaries at positions 1, 4, and 7 using slicing.Use start, stop, and step together
print(salary.iloc[1:8:3])
```
