```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ]
)

print(salary)
```

```python
#Display the salary of the first employee.Position 0
print(salary.iloc[0])
```

```python
#Display the salary of the third employee.Position 2
print(salary.iloc[2])
```

```python
#Display the salary of Tayab.
#Find Rahul's integer position.
print(salary.iloc[[6]])

```

```python
#Display the salary of Mirzan.Last position
print(salary.iloc[[-1]])
```

```python
#Display the salary of Suhana.Position 1
print(salary.iloc[[1]])
```

```python
#Display the salary of the sixth employee.Position 5
print(salary.iloc[[5]])
```

```python
#Display the salary of the employee at position 7.Remember, positions start at 0.
print(salary.iloc[[6]])
```

```python
#Display the salary of the last employee using negative indexing.Use a negative integer.
print(salary.iloc[[-1]])
```

```python
#Display the salary of the second-last employee.Use a negative integer.
print(salary.iloc[[-2]])
```

```python
#Display the salary of the employee at position 4
print(salary.iloc[[4]])
```
