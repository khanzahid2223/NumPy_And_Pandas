```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ]
)
```

```python
#Equal
salary.loc[salary.eq(30000)]
```

```python
#Not Equal
salary.loc[salary.ne(30000)]
```

```python
#Greater Than
salary.loc[salary.gt(35000)]
```

```python
#Greater than or equal
salary.loc[salary.ge(45000)]
```

```python
#Less Than
salary.loc[salary.lt(45000)]
```

```python
#Less Than or Equal
salary.loc[salary.le(50000)]
```
