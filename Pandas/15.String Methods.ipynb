```python
import pandas as pd
```

```python
### Q1. Convert Names to Uppercase
#Convert all names to uppercase.
names = pd.Series(["zahid", "rahul", "priya", "aman"])
print(names.str.upper())
```

```python
### Q2. Convert Product Names to Lowercase
#Convert all product names to lowercase.
products = pd.Series(["LAPTOP", "MOUSE", "KEYBOARD", "MONITOR"])
print(products.str.lower())
```

```python
### Q3. Convert Names to Title Case
#Convert each customer's name to title case.
customers = pd.Series(["rahul kumar", "zahid khan", "priya sharma"])
print(customers.str.title())
```

```python
### Q4. Remove Extra Spaces
#Remove leading and trailing spaces from every city.
cities = pd.Series([" Bengaluru ", " Chennai", "Mumbai ", " Hyderabad "])
print(cities)
print(cities.str.strip())
```

```python
### Q5. Count Characters
#Count the number of characters in each department name.
departments = pd.Series(["Sales", "Finance", "IT", "Marketing"])
print(departments.str.len())
```

```python
### Q6. Check for a Keyword
#Check which email addresses contain "gmail".
emails = pd.Series([
    "zahid@gmail.com",
    "rahul@yahoo.com",
    "priya@gmail.com",
    "aman@outlook.com"
])
print(emails.str.contains("gmail"))
print(emails.loc[emails.str.contains("gmail")])
```

```python
### Q7. Check Starting Characters
#Check which product names start with "Laptop".
products = pd.Series(["Laptop Pro", "Mouse Wireless", "Laptop Air", "Keyboard"])
print(products.str.startswith("Laptop"))
print(products.loc[products.str.startswith("Laptop")])
```

```python
### Q8. Check Ending Characters
#Identify the files that end with ".csv".
files = pd.Series(["report.csv", "image.png", "data.csv", "notes.txt"])
print(files.str.endswith(".csv"))
print(files.loc[files.str.endswith(".csv")])
```

```python
### Q9. Find a Word
#Check which feedback entries contain "excellent"
feedback = pd.Series([
    "excellent service",
    "poor delivery",
    "excellent quality",
    "average service"
])
print(feedback.str.contains("excellent"))
print(feedback.loc[feedback.str.contains("excellent")])
```

```python
### Q10. Case-Insensitive Search
#Find all entries that match "apple", regardless of capitalization.
brands = pd.Series(["Apple", "APPLE", "apple", "Samsung", "Dell"])
print(brands.loc[brands.str.lower()=="apple"])

```

```python
### Q11. Replace Text
#Replace "Department" with "Dept" in every entry.
departments = pd.Series([
    "IT Department",
    "HR Department",
    "IT Department",
    "Sales Department"
])
print(departments.str.replace("Department","Dept"))
```

```python
### Q12. Split Full Names
#Split each full name into first name and last name.
names = pd.Series(["Zahid Khan", "Rahul Sharma", "Priya Reddy"])
print(names.str.split(" "))
```

```python
### Q13. Extract Domain Names
#Extract only the domain portion, such as "gmail.com".
emails = pd.Series([
    "zahid@gmail.com",
    "rahul@yahoo.com",
    "priya@outlook.com"
])
print(emails.loc[emails.str.contains("gmail.com")])
```

```python
### Q14. Remove a Prefix
#Remove the "EMP" prefix from every employee ID.
employee_ids = pd.Series(["EMP101", "EMP102", "EMP103", "EMP104"])
print(employee_ids.str.removeprefix("EMP"))
```

```python
### Q15. Pad Employee IDs
#Make every ID exactly five characters long by adding zeros to the left.
ids = pd.Series(["1", "12", "123", "1234"])
print(ids.str.zfill(5))
```

```python
### Q16. Count a Specific Character
#Count how many times the character "a" appears in each email address.
emails = pd.Series([
    "a@gmail.com",
    "rahul@yahoo.com",
    "priya@outlook.com"
])
print(emails.str.count("a"))
```

```python
### Q17. Extract Digits from Product Codes
#Extract only the numeric portions from each product code.
codes = pd.Series(["PROD101", "ITEM205", "PROD310", "ITEM420"])
print(codes.str.slice(-3))
```

```python
### Q18. Validate Email Format
#Create a Boolean Series identifying entries with text before "@", a domain name, and a ".com" ending.
emails = pd.Series([
    "zahid@gmail.com",
    "rahul.yahoo.com",
    "priya@outlook.com",
    "aman@gmail"
])
result = emails.str.contains("@") & emails.str.endswith(".com")

print(result)
print(emails.loc[emails.str.contains("@") & emails.str.endswith(".com")])
```

```python
### Q19. Anonymize Phone Numbers
#Replace the first six digits with "X", leaving only the last four digits visible.
phones = pd.Series([
    "9876543210",
    "8765432109",
    "7654321098"
])
print("XXXXXX"+phones.str.slice(6))
```

```python
### Q20. Customer Data Cleaning Challenge
#Perform all three operations:
customers = pd.Series([
    "  ZAHID KHAN  ",
    "rahul sharma",
    " PRIYA REDDY",
    "AMAN KUMAR "
])
#1. Remove leading and trailing spaces.
#2. Convert names to title case.
#3. Check which cleaned names contain "Kumar".
Cleaned=customers.str.strip()
print(Cleaned)
Cleaned=customers.str.title()
print(Cleaned)
print(Cleaned.str.contains("Kumar"))
```
