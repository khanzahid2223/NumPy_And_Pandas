```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ],
    name="Salary"
)

print(salary)
```

```python
#Reset the index of the salary Series and convert it into a DataFrame.Hint: Use reset_index() with to_frame().
salary.reset_index()
```

```python
#Reset the index and store the result in a new variable called df.Hint: Assign the result to df.
df=salary.reset_index()
df
```

```python
#Reset the index and keep the original Series unchanged.Hint: reset_index() returns a new object by default.
new_salary=salary.reset_index()
new_salary
```

```python
#Reset the index and rename the columns to Employee and Salary.Hint: Convert to a DataFrame, then rename the columns.
salary.reset_index().rename(columns={"index":"Employee","Salary":"Salary"})
```

```python
#Reset the index and make Employee the default numerical index.Hint: The default index starts at 0.
salary.reset_index(drop=True)
```

```python
#Reset the index, then display only the first five rows.Hint: Use head(5).
new_salary=salary.reset_index()
print(new_salary)
print(new_salary[0:6])
```
