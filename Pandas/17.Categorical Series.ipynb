```python
import pandas as pd
import numpy as np
```

```python
### Q1. Employee Department Analysis
# → Count the number of employees in each department using `value_counts()`.
# → Find the department with the highest number of employees.
# → Find the total number of recorded employees.
departments = pd.Series(["IT", "HR", "IT", "Finance", "HR", "IT", "Sales", "Finance", "IT", "Sales"])
print(departments.value_counts())
print(departments.value_counts().idxmax())
print(departments.count())
```

```python
### Q2. Customer Feedback Analysis
# → Find the most common feedback.
# → Find how many customers gave "Excellent" feedback.
# → Count each feedback category.
feedback = pd.Series(["Good", "Excellent", "Good", "Average", "Poor", "Excellent", "Good", "Poor", "Good", "Average"])
print(feedback.value_counts().idxmax())
print(feedback.value_counts()["Excellent"])
print(feedback.count())
```

```python
### Q3. Product Category Sales
# → Count products in each category.
# → Find the most frequently occurring category.
# → Count how many products belong to "Clothing".
categories = pd.Series(["Electronics", "Clothing", "Electronics", "Grocery", "Clothing", "Electronics", "Grocery", "Clothing", "Electronics"])
print(categories.value_counts())
print(categories.mode())
print(categories.value_counts()["Clothing"])
```

```python
### Q4. Student Grade Analysis
# → Count students in each grade.
# → Find the most common grade.
# → Count students who received grade "A".
grades = pd.Series(["A", "B", "A", "C", "B", "A", "D", "B", "A", "C", "A"])
print(grades.value_counts())
print(grades.value_counts().idxmax())
print(grades.value_counts()["A"])
```

```python
### Q5. Payment Method Analysis
# → Count transactions by payment method.
# → Find the most frequently used payment method.
# → Count transactions made using UPI.
payments = pd.Series(["UPI", "Card", "Cash", "UPI", "UPI", "Card", "Cash", "UPI", "Card", "UPI"])
print(payments.value_counts())
print(payments.mode())
print(payments.value_counts()["UPI"])
```

```python
### Q6. Website Device Analysis
# → Count visits by device type.
# → Find the most common device type.
# → Count visits from Desktop devices.
devices = pd.Series(["Mobile", "Desktop", "Mobile", "Tablet", "Mobile", "Desktop", "Mobile", "Tablet", "Desktop", "Mobile"])
print(devices.value_counts())
print(devices.mode())
print(devices.value_counts()["Desktop"])
print(devices.value_counts().sort_index())
```

```python
### Q7. Employee Employment Status
# → Count employees by employment status.
# → Find the most common status.
# → Count employees who are Inactive.
status = pd.Series(["Active", "Inactive", "Active", "Active", "On Leave", "Inactive", "Active", "On Leave", "Active"])
print(status.value_counts())
print(status.mode())
print(status.value_counts()["Inactive"])
```

```python

### Q8. Order Delivery Status
# → Count orders by delivery status.
# → Find the most common delivery status.
# → Count cancelled orders.
delivery = pd.Series(["Delivered", "Pending", "Cancelled", "Delivered", "Delivered", "Pending", "Returned", "Delivered", "Cancelled", "Delivered"])
print(delivery.value_counts())
print(delivery.mode())
print(delivery.value_counts()["Cancelled"])
print((delivery.value_counts(normalize=True)*100).round(2))
print(delivery.describe())
```
