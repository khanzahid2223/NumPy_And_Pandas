```python
import numpy as np
import pandas as pd
```

```python
### Q1. Employee Salary Analysis
# → Find the number of missing salary values.
# → Count the non-missing salary values.
# → Calculate the average salary, ignoring missing values.
salary = pd.Series([25000, 30000, np.nan, 45000, 35000, np.nan, 50000])
print(salary.isna().sum())
print(salary.count())
print(salary.mean())
```

```python
### Q2. Website Traffic Analysis
# → Identify which values are missing.
# → Count the missing values.
# → Calculate the total recorded visitors.
visitors = pd.Series([1200, np.nan, 1800, 2200, np.nan, 2500, 1900])
print(visitors[visitors.isna()])
print(visitors.isna().sum())
print(visitors.value_counts())
```

```python
# ### Q3. Student Marks Analysis
# → Count the missing marks.
# → Calculate the average marks, ignoring missing values.
# → Replace missing marks with 0 in a new Series.
marks = pd.Series([85, 90, np.nan, 76, 88, np.nan, 95, 79])
print(marks.isna().sum())
print(marks.mean())
print(marks.fillna(0))
```

```python
### Q4. Product Price Analysis
# → Find the missing values using `isna()`.
# → Count the missing values using `isna().sum()`.
# → Fill missing prices with the mean price.
prices = pd.Series([499, 799, np.nan, 1299, 999, np.nan, 1499])
print(prices.isna())
print(prices.isna().sum())
print(prices.fillna(round(prices.mean(),2)))
```

```python
### Q5. Customer Age Analysis
# → Count the missing ages.
# → Find the median age, ignoring missing values.
# → Fill missing ages with the median age.
ages = pd.Series([22, 35, np.nan, 28, 41, np.nan, 30, 26])
print(ages.isna().sum())
print(ages.median())
print(ages.fillna(ages.median()))
```

```python
### Q6. Monthly Sales Analysis
# → Calculate the number of recorded sales values.
# → Find the total sales, ignoring missing values.
# → Replace missing sales values with 0.
sales = pd.Series([12000, 15000, np.nan, 18000, 22000, np.nan, 25000])
print(sales.count())
print(sales.sum())
print(sales.fillna(0))
```

```python
### Q7. Customer Ratings Analysis
# → Count the missing ratings.
# → Find the maximum and minimum ratings.
# → Drop missing ratings and store the result in a new Series.
ratings = pd.Series([4.5, 3.8, np.nan, 4.9, 4.2, np.nan, 3.5])
print(ratings.isna().sum())
print(ratings.min())
print(ratings.max())
print(ratings.dropna().reset_index(drop=True))
```

```python
### Q8. Employee Experience Analysis
# → Check whether each value is missing.
# → Count non-missing values.
# → Fill missing experience values with the median experience.
experience = pd.Series([1, 3, np.nan, 5, 2, np.nan, 7, 4])
print(experience.isna().sum())
print(experience.value_counts())
print(experience.fillna(experience.mean()))
```

```python
### Q9. Delivery Time Analysis
# → Count the missing delivery times.
# → Calculate the average delivery time, ignoring missing values.
# → Fill missing delivery times with the mean delivery time.
delivery_time = pd.Series([30, 45, np.nan, 60, 35, np.nan, 50, 40])
print(delivery_time.isna().sum())
print(delivery_time.mean())
print(delivery_time.fillna(delivery_time.mean()))
```

```python
### Q10. Business Revenue Analysis
# → Find the total recorded revenue.
# → Calculate the average recorded revenue.
# → Count missing revenue values.
# → Drop missing values and calculate the new Series length.
revenue = pd.Series([50000, np.nan, 75000, 60000, np.nan, 90000, 85000],index=["A","B","C","D","E","F","G"])
print(revenue.sum())
print(revenue.mean())
print(revenue.isna().sum())
print(revenue.dropna().reset_index())
```
