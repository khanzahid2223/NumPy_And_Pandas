```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ],
    name="Salary"
)

print(salary)
```

```python
#Filter employees whose salary is greater than ₹35,000.Hint: Use salary > 35000 inside salary[...].
result=salary.loc[salary>35000]
print(result)
```

```python
#Filter employees whose salary is less than ₹30,000.Hint: Use the less-than operator.
result=salary.loc[salary<30000]
print(result)
```

```python
#Filter employees whose salary is between ₹30,000 and ₹45,000, inclusive.Hint: Combine two conditions using & and put each condition in parentheses.
result=salary[salary.between(30000,45000)]
print(result)
```

```python
#. Filter employees whose salary is exactly ₹50,000.Hint: Use the equality operator ==.
result=salary.loc[salary==50000]
print(result)
```

```python
# Filter employees whose salary is either below ₹30,000 or above ₹45,000.Hint: Combine conditions using | and put each condition in parentheses.
result=salary.loc[(salary<30000) | (salary>45000)]
print(result)
```

```python
mask=salary>40000
print(mask)
print(f"Sum:{mask.sum()}")
print(f"Any:{mask.any()}")
print(f"All:{mask.all()}")
print(f"Length:{len(salary.loc[mask])}")
```
