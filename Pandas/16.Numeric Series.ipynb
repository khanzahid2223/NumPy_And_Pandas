```python
import pandas as pd
import numpy as np
```

```python
### Q1. Employee Salary Analysis
#Calculate the total salary.
#Calculate the average salary.
#Find the minimum and maximum salary.
#Count the available salary values.
salary = pd.Series([25000, 32000, 28000, np.nan, 45000, 38000, np.nan])
print(salary.sum())
print(salary.mean())
print(salary.min())
print(salary.max())
print(salary.count())
```

```python
### Q2. Monthly Sales Analysis
# → Calculate total sales.
# → Calculate average sales.
# → Find the highest and lowest sales.
# → Count the missing values.
sales = pd.Series([12000, 15000, np.nan, 18000, 22000, 14000, np.nan, 25000])
print(sales.sum())
print(sales.mean())
print(sales.min())
print(sales.max())
print(sales.isna().sum())
```

```python
### Q3. Product Price Analysis
# → Calculate the average product price.
# → Find the minimum and maximum price.
# → Calculate the total of all available prices.
prices = pd.Series([499, 799, np.nan, 1299, 999, 599, np.nan, 1499])
print(prices.mean())
print(prices.min())
print(prices.max())
print(prices.sum())
```

```python
### Q4. Customer Ratings
# → Calculate the average rating.
# → Find the highest and lowest rating.
# → Count the available ratings.
ratings = pd.Series([4.5, 3.8, np.nan, 4.9, 3.5, 4.2, np.nan, 4.7])
print(ratings.mean())
print(ratings.min())
print(ratings.max())
print(ratings.count())
```

```python
### Q5. Website Traffic
# → Calculate the total number of recorded visitors.
# → Calculate the average daily visitors.
# → Find the maximum and minimum visitor counts.
visitors = pd.Series([1200, 1500, 1800, np.nan, 2200, 1900, np.nan, 2500])
print(visitors.sum())
print(visitors.mean())
print(visitors.min())
print(visitors.max())
```

```python

### Q6. Employee Performance Scores
# → Calculate the average score.
# → Calculate the total score.
# → Find the highest and lowest score.
# → Count the number of valid scores.
scores = pd.Series([85, 92, np.nan, 78, 88, 95, 67, np.nan, 90])
print(scores.mean())
print(scores.sum())
print(scores.min())
print(scores.max())
print(scores.count())
```

```python
# ### Q7. Delivery Time Analysis
# → Calculate the average delivery time.
# → Find the longest and shortest delivery times.
# → Calculate the total recorded delivery hours.
delivery_hours = pd.Series([24, 36, np.nan, 48, 30, 18, np.nan, 42])
print(delivery_hours.mean())
print(delivery_hours.min())
print(delivery_hours.max())
print(delivery_hours.count())
```

```python
scores = pd.Series([85, 92, np.nan, 78, 88, 95, 67, np.nan, 90])
print(scores.std())
print(scores.var())
print(scores.quantile(0.25))
print(scores.idxmax())
print(scores.idxmin())
print(scores.max()-scores.min())
print(scores.describe())
print(scores.fillna(scores.mean()))
print(scores.dropna())
```
