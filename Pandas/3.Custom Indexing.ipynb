```python
import pandas as pd
```

```python
#Student Marks
#Create a Series with marks [85, 90, 78, 95] and custom indexes ['Maths', 'Python', 'SQL', 'AI'].
Marks=[85, 90, 78, 95]
Subjects=['Maths', 'Python', 'SQL', 'AI']
Marks_Series=pd.Series(Marks,index=Subjects)
Marks_Series
```

```python
#Employee Salaries
#Create a Series with salaries [25000, 30000, 28000, 35000] and indexes ['E101', 'E102', 'E103', 'E104'].
Salary=[25000, 30000, 28000, 35000]
Employee=['E101', 'E102', 'E103', 'E104']
Employee_Series=pd.Series(Salary,index=Employee)
Employee_Series
```

```python
#Product Prices
#Create a Series with prices [100, 40, 60, 150] and indexes ['Apple', 'Banana', 'Orange', 'Mango'].
Price=[100, 40, 60, 150]
Fruits=['Apple', 'Banana', 'Orange', 'Mango']
Fruits_Series=pd.Series(Price,index=Fruits)
Fruits_Series
```

```python
# Daily Sales
#Create a Series with sales [1200, 1500, 1800, 1300, 2000] and indexes ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'].
Sales=[1200, 1500, 1800, 1300, 2000]
Days=['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
Sales_Series=pd.Series(Sales,index=Days)
Sales_Series
```

```python
#City Population
#Create a Series with populations in lakhs [85, 120, 70, 95] and indexes ['Bengaluru', 'Mumbai', 'Chennai', 'Hyderabad'].
Population=[85, 120, 70, 95]
City=['Bengaluru', 'Mumbai', 'Chennai', 'Hyderabad']
Population_Series=pd.Series(Population,index=City)
Population_Series
```

```python
#Subject Attendance
#Create a Series with attendance percentages [90, 85, 95, 80] and indexes ['Python', 'SQL', 'Maths', 'AI']. Then access the attendance for SQL.
Attendance_Percentages=[90, 85, 95, 80]
Subjects=['Python', 'SQL', 'Maths', 'AI']
Attendance_Percentages_Series=pd.Series(Attendance_Percentages,index=Subjects)
print(Attendance_Percentages_Series)
print(Attendance_Percentages_Series["SQL"])
```

```python
# Monthly Expenses
#Create a Series with expenses [5000, 3500, 2000, 4500, 3000] and indexes ['Rent', 'Food', 'Travel', 'Shopping', 'Bills']. Then display only Food and Bills.
Expenses=[5000, 3500, 2000, 4500, 3000]
items=['Rent', 'Food', 'Travel', 'Shopping', 'Bills']
Expenses_Series=pd.Series(Expenses,index=items)
print(Expenses_Series)
print(Expenses_Series[["Food","Bills"]])
```

```python
#Cricket Scores
#Create a Series with scores [45, 78, 32, 91, 60] and indexes ['Rohit', 'Virat', 'Gill', 'Rahul', 'Surya']. Access Virat's and Rahul's scores.
Score=[45, 78, 32, 91, 60]
players=['Rohit', 'Virat', 'Gill', 'Rahul', 'Surya']
Score_Series=pd.Series(Score,index=players)
print(Score_Series)
print(Score_Series[["Virat","Rahul"]])
```

```python
#Product Stock
#Create a Series with stock quantities [50, 20, 75, 10] and indexes ['Laptop', 'Mouse', 'Keyboard', 'Monitor']. Change the stock of Mouse to 35 and print the updated Series.
Stock_Quantities=[50, 20, 75, 10]
Prodects=['Laptop', 'Mouse', 'Keyboard', 'Monitor']
Stock_Quantities_Series=pd.Series(Stock_Quantities,index=Prodects)
print(Stock_Quantities_Series)
Stock_Quantities_Series["Mouse"]=35
print(Stock_Quantities_Series)
```

```python
# Real-World Sales Challenge
#Create a Series with sales [10000, 15000, 8000, 20000, 12000] and indexes ['Store_A', 'Store_B', 'Store_C', 'Store_D', 'Store_E']. Then:
# Access the sales of Store_D.
#Display Store_A and Store_C.
#Update Store_B sales to 18000
Sales=[10000, 15000, 8000, 20000, 12000]
Store=['Store_A', 'Store_B', 'Store_C', 'Store_D', 'Store_E']
Sales_Series=pd.Series(Sales,index=Store)
print(Sales_Series)
print(Sales_Series["Store_D"])
print(Sales_Series[["Store_A","Store_C"]])
Sales_Series["Store_B"]=18000
print(Sales_Series)
```
