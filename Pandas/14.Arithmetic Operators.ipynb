```python
import pandas as pd
```

```python
#Increase Employee Salaries
#Create a Series with salaries [25000, 30000, 35000, 40000] and indexes ['E101', 'E102', 'E103', 'E104'].
#Add ₹5,000 to every employee's salary.
Salaries=[25000, 30000, 35000, 40000]
Employee_Salaries=pd.Series(Salaries,index=['E101', 'E102', 'E103', 'E104'])
Employee_Salaries.add(5000)
```

```python
#Apply a Discount
#Create a Series with product prices [1000, 2000, 1500, 3000] and indexes ['Laptop_Bag', 'Keyboard', 'Mouse', 'Headphones'].
#Reduce every price by ₹100
Price=[1000, 2000, 1500, 3000]
Discount=pd.Series(Price,index=['Laptop_Bag', 'Keyboard', 'Mouse', 'Headphones'])
Discount.sub(100)
```

```python
#Calculate Total Sales
#Create a Series with daily sales [1200, 1500, 1800, 2000, 2500].
#Add ₹500 to every day's sales and display the updated Series.
Sales=[1200, 1500, 1800, 2000, 2500]
Total_Sales=pd.Series(Sales)
Total_Sales.add(500)
```

```python
#Calculate Annual Salary
#Create a Series with monthly salaries [25000, 30000, 40000, 50000].
#Multiply every value by 12 to calculate annual salaries.
Salaries=[25000, 30000, 40000, 50000]
Annual_Salaries=pd.Series(Salaries)
Annual_Salaries.mul(12)
```

```python
#Calculate Price After Discount
#Create a Series with prices [1000, 2500, 4000, 5000].
#Apply a 10% discount to every product and display the final prices.
Price=[1000, 2500, 4000, 5000]
Price_Discount=pd.Series(Price)
Price_Discount.mul(1.10)
```

```python
# Calculate Average Marks
#Create a Series with student marks [85, 90, 78, 95, 82].
#Use an arithmetic or aggregation method to calculate the average marks.
Marks=[85, 90, 78, 95, 82]
Averege_Marks=pd.Series(Marks)

print(Averege_Marks.sum()/len(Marks))
```

```python
#Calculate Remaining Stock
#Create a Series with stock quantities [100, 80, 150, 60] and another Series with sold quantities [20, 15, 50, 10], using matching product indexes.
#Subtract the sold quantities from the stock quantities.
Stock=pd.Series([100, 80, 150, 60])
Sold_Quantities=pd.Series([20, 15, 50, 10])
Remaining=Stock-Sold_Quantities
Remaining
```

```python
# Compare Two Stores
#Create two Series with matching indexes ['Monday', 'Tuesday', 'Wednesday', 'Thursday'].
#Store A sales: [1000, 1500, 1200, 1800]
#Store B sales: [800, 1700, 1000, 2000]
#Subtract Store B sales from Store A sales to find the daily sales differences.
StoreA=pd.Series([1000, 1500, 1200, 1800],index=['Monday', 'Tuesday', 'Wednesday', 'Thursday'])
StoreB=pd.Series([800, 1700, 1000, 2000],index=['Monday', 'Tuesday', 'Wednesday', 'Thursday'])
Daily_Sales=StoreA-StoreB
Daily_Sales
```

```python
#Calculate Electricity Bills
#Create a Series with electricity consumption [100, 150, 200, 250] in units.
#Assume each unit costs ₹7. Multiply every value by 7 to calculate each bill.
Electricity_Consumption=pd.Series([100, 150, 200, 250])
Cost=Electricity_Consumption.mul(7)
Cost
```

```python
#Business Revenue Analysis
#Create two Series with indexes ['Branch_A', 'Branch_B', 'Branch_C', 'Branch_D'].
#Revenue: [50000, 75000, 60000, 90000]
#Expenses: [30000, 45000, 25000, 50000]
#Calculate profit using subtraction.
#Calculate profit margin as a percentage of revenue.
#Increase every branch's revenue by 15%
Revenue=pd.Series([50000, 75000, 60000, 90000],index=['Branch_A', 'Branch_B', 'Branch_C', 'Branch_D'])
Expenses=pd.Series([30000, 45000, 25000, 50000],index=['Branch_A', 'Branch_B', 'Branch_C', 'Branch_D'])
Profit=Revenue-Expenses
print(Profit)
Profit_Margin=(Profit/Revenue)*100
print(Profit_Margin)
Revenue_Inc=Revenue.mul(1.50)
Revenue_Inc
```
