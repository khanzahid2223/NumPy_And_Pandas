```python
import pandas as pd

salary = pd.Series(
    [25000, 32000, 28000, 45000, 38000,
     50000, 30000, 42000, 35000, 60000],
    index=[
        "Zahid", "Suhana", "Arosh", "Furkhan", "Chandini",
        "Asfiya", "Tayab", "Faizal", "Sufiyan", "Mirzan"
    ]
)
```

```python
#Check whether the salary value 25000 exists in the Series.Hint: Use the `in` operator with `salary.values`.
print(25000 in salary.values)
```

```python
#Check whether the employee name 'Zahid' exists in the Series index.Hint: Use `salary.index`.
salary.loc[salary.index.isin(["Zahid"])]
```

```python
#Check whether the salary value 40000 exists in the Series.Hint: Test membership in the values.
print(40000 in salary.values)
```

```python
#Check whether 'Mirzan' exists as an index label.Hint: Use the `in` operator.
salary.loc[salary.index.isin(["Mirzan"])]
```

```python
#Check whether the salary value 60000 exists in the Series.Hint: Test membership in `salary.values`.
print(60000 in salary.values)
```
