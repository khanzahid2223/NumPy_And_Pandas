```python
import pandas as pd
import numpy as np
```

```python
### Q1. Employee Salary Analysis
# → Keep salaries greater than or equal to ₹30,000 unchanged.
# → Replace salaries below ₹30,000 with `0`.
# → Store the result in a new Series.
salary = pd.Series([25000, 45000, 18000, 60000, 32000, 15000, 75000])
s=salary.where(salary>=30000,0)
s
```

```python
### Q2. Student Marks Analysis
# → Keep marks greater than or equal to 40 unchanged.
# → Replace marks below 40 with `"Fail"`.
# → Store the result in a new Series.
marks = pd.Series([85, 42, 91, 35, 76, 28, 68, 95])
m=marks.where(marks>=40,"Fail")
m
```

```python
### Q3. E-commerce Product Discount
# → Keep prices greater than ₹1,000 unchanged.
# → Replace prices less than or equal to ₹1,000 with `NaN`.
# → Store the result in a new Series.
prices = pd.Series([500, 1500, 800, 2500, 1200, 400, 3000])
p=prices.where(prices>=1000,np.nan)
p
```

```python
### Q4. Website Traffic Analysis
# → Keep visitor counts greater than or equal to 1,000 unchanged.
# → Replace counts below 1,000 with `0`.
# → Store the result in a new Series.
visitors = pd.Series([500, 1200, 800, 3500, 2200, 400, 5000])
v=visitors.where(visitors>=1000,0)
v
```
